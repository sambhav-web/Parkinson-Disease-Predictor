## We can simply remove this name columns.

```python
df.drop('name',axis=1,inplace=True)
```

```python
x=df.drop('status',axis=1)
y=df['status']
```

```python
y.value_counts()
```

## 1 -> Parkinson's Positive
## 0 -> Healthy
## As number of 1 occurs 3 times the number of healthy people.Data is highly imbalanced.

```python
df.groupby('status').mean()
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
x_train
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import BernoulliNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.utils.class_weight import compute_class_weight
classes=df['status'].unique()
# weights=compute_class_weight(class_weight='balanced',classes=classes,y=y_train)
weights=[5,8]
class_weights=dict(zip(classes,weights))
print(class_weights)
```

```python
accuracy={}
models={
    'LogisticRegression':LogisticRegression(class_weight=class_weights),
    'BernoulliNB':BernoulliNB(),
    'SVC':SVC(class_weight=class_weights)
}
for i in range(len(list(models))):
  m={}
  model=list(models.values())[i]
  model.fit(x_train,y_train)
  y_train_pred=model.predict(x_train)
  y_test_pred=model.predict(x_test)
  m["Train_Acc"]=accuracy_score(y_train,y_train_pred)
  m["Test_Acc"]=accuracy_score(y_test,y_test_pred)
  accuracy[list(models)[i]]=m
  print(f'For {list(models)[i]}')
  print('Training Report :\n',classification_report(y_train,y_train_pred))
  print('Testing Report :\n',classification_report(y_test,y_test_pred))
print(pd.DataFrame(accuracy))
```

```python
model=SVC(class_weight=class_weights)
model.fit(x_train,y_train)
y_train_pred=model.predict(x_train)
y_test_pred=model.predict(x_test)
```

```python
confusion_matrix(y_train,y_train_pred)
```

```python
confusion_matrix(y_test,y_test_pred)
```

```python
print('Train Acc : ',accuracy_score(y_train,y_train_pred))
print('Test Acc : ',accuracy_score(y_test,y_test_pred))
```

```python
import pickle

# Save the trained SVC model and the fitted scaler
with open('parkinson_model.pkl', 'wb') as model_file:
    pickle.dump(model, model_file)

with open('scaler.pkl', 'wb') as scaler_file:
    pickle.dump(scaler, scaler_file)

print("Model and Scaler successfully saved!")
```
